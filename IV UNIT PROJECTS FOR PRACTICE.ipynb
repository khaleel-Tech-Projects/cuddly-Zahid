```python
import numpy as np
import pandas as pd
import seaborn as sns

def get_tips_data():
    try:
        df = sns.load_dataset('tips')
        if df.empty or len(df) == 0:
            raise ValueError("Empty dataset fetched.")
        return df
    except Exception:
        # Standalone offline fallback matching Seaborn's tips schema exactly
        np.random.seed(42)
        n = 244
        bills = np.round(np.random.gamma(shape=9, scale=2.2, size=n), 2)
        tips = np.round(bills * np.random.uniform(0.10, 0.22, size=n), 2)
        return pd.DataFrame({
            'total_bill': bills,
            'tip': tips,
            'sex': np.random.choice(['Male', 'Female'], size=n, p=[0.64, 0.36]),
            'smoker': np.random.choice(['No', 'Yes'], size=n, p=[0.62, 0.38]),
            'day': np.random.choice(['Thur', 'Fri', 'Sat', 'Sun'], size=n, p=[0.25, 0.08, 0.36, 0.31]),
            'time': np.random.choice(['Dinner', 'Lunch'], size=n, p=[0.72, 0.28]),
            'size': np.random.choice([1, 2, 3, 4, 5, 6], size=n, p=[0.02, 0.64, 0.16, 0.15, 0.02, 0.01])
        })
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

np.random.seed(42)
n = 244
bills = np.round(np.random.gamma(shape=9, scale=2.2, size=n), 2)
df = pd.DataFrame({
    'total_bill': bills,
    'tip': np.round(bills * np.random.uniform(0.10, 0.22, size=n), 2),
    'time': np.random.choice(['Lunch', 'Dinner'], size=n, p=[0.3, 0.7])
})

sns.set_theme(style='whitegrid', font_scale=1.1)

fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    style='time',
    palette={'Lunch': '#2ecc71', 'Dinner': '#e74c3c'},
    s=80,
    alpha=0.85,
    ax=ax
)

ax.set_title('Tip vs. Total Bill by Meal Timing', fontsize=15, weight='bold', pad=15)
ax.set_xlabel('Total Bill ($)', fontsize=12, weight='semibold')
ax.set_ylabel('Tip Amount ($)', fontsize=12, weight='semibold')
ax.set_xlim(0, 60)
ax.set_ylim(0, 15)

ax.legend(title='Dining Period', title_fontsize='11', loc='upper left', frameon=True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

np.random.seed(42)
n = 244
bills = np.round(np.random.gamma(shape=9, scale=2.2, size=n), 2)
df = pd.DataFrame({
    'total_bill': bills,
    'tip': np.round(bills * np.random.uniform(0.10, 0.22, size=n), 2),
    'day': np.random.choice(['Thur', 'Fri', 'Sat', 'Sun'], size=n),
    'time': np.random.choice(['Lunch', 'Dinner'], size=n)
})

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Subplot 1: Total bill distribution
sns.histplot(df['total_bill'], kde=True, ax=axes[0, 0], color='teal')
axes[0, 0].set_title('Total Bill Distribution')

# Subplot 2: Average Tip by Day
sns.barplot(data=df, x='day', y='tip', ax=axes[0, 1], errorbar=None, palette='Blues')
axes[0, 1].set_title('Average Tip by Day')

# Subplot 3: Bill vs Tip regression
sns.regplot(data=df, x='total_bill', y='tip', ax=axes[1, 0], scatter_kws={'alpha': 0.5})
axes[1, 0].set_title('Bill vs Tip Linear Relationship')

# Subplot 4: Customer counts by time
sns.countplot(data=df, x='time', ax=axes[1, 1], palette='Set1')
axes[1, 1].set_title('Traffic Count by Service Time')

fig.suptitle('Exploratory Data Analysis Dashboard', fontsize=16, weight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

try:
    from sklearn.datasets import load_iris
    iris = load_iris(as_frame=True)
    df = iris.frame
    df['species'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
    df = df.drop(columns=['target'])
except Exception:
    import numpy as np
    df = pd.DataFrame(np.random.rand(150, 3), columns=['feat_1', 'feat_2', 'feat_3'])
    df['species'] = np.random.choice(['A', 'B', 'C'], size=150)

g = sns.PairGrid(df, hue='species', palette='Dark2')

g.map_diag(sns.kdeplot, fill=True, common_norm=False, alpha=0.4)
g.map_upper(sns.scatterplot, alpha=0.7)
g.map_lower(sns.kdeplot, levels=4, alpha=0.6)

g.add_legend(title='Species')
g.fig.subplots_adjust(top=0.94)
g.fig.suptitle('PairGrid Diagnostics', fontsize=14, weight='bold')

plt.show()
```

```python
import plotly.express as px
import numpy as np
import pandas as pd

# Creating local DataFrame to avoid px.data internet fetching issues
np.random.seed(42)
n = 244
bills = np.round(np.random.gamma(shape=9, scale=2.2, size=n), 2)
df = pd.DataFrame({
    'total_bill': bills,
    'tip': np.round(bills * np.random.uniform(0.10, 0.22, size=n), 2),
    'day': np.random.choice(['Thur', 'Fri', 'Sat', 'Sun'], size=n),
    'size': np.random.choice([1, 2, 3, 4, 5, 6], size=n),
    'sex': np.random.choice(['Male', 'Female'], size=n),
    'smoker': np.random.choice(['Yes', 'No'], size=n),
    'time': np.random.choice(['Lunch', 'Dinner'], size=n)
})

fig = px.scatter(
    df,
    x='total_bill',
    y='tip',
    color='day',
    size='size',
    hover_data=['sex', 'smoker', 'time'],
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip Amount ($)'},
    title='Interactive EDA: Bill vs. Tip (Offline Supported)',
    template='plotly_white'
)

fig.update_traces(marker=dict(opacity=0.8, line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(legend_title_text='Day of Week')

fig.show()
```
